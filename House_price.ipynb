```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.svm import SVR
from sklearn.ensemble import RandomForestRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
```

```python
data=pd.read_csv("house_prices_clean.csv")
```

```python
data_khoob=data[data['Price']>100000],data[data['Age']>18]
```

```python
x=data.drop(["Price"],axis=1)
y=data["Price"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model_lin=LinearRegression()
model_lin.fit(x_train,y_train)
pishbini_model_lin=model_lin.predict(x_test)
```

```python
deghat_lin=r2_score(pishbini_model_lin,y_test)#این
round(deghat_lin,2)
```

```python
model_svm=SVR()
model_svm.fit(x_train,y_train)
y_pishbini_svm=model_svm.predict(x_test)
```

```python
deghat_svm=r2_score(y_test,y_pishbini_svm)
round(deghat_svm,2)
```

```python
model_random=RandomForestRegressor()
model_random.fit(x_train,y_train)
y_pishbini_random=model_random.predict(x_test)
```

```python
deghat_random=r2_score(y_test,y_pishbini_random)
round(deghat_random,2)
```

```python
model_kn=KNeighborsRegressor()
model_kn.fit(x_train,y_train)
y_pishbini_kn=model_kn.predict(x_test)
```

```python
deghat_kn=r2_score(y_test,y_pishbini_kn)
round(deghat_kn,2)
```

```python
model_deci=DecisionTreeRegressor(random_state=42)
model_deci.fit(x_train,y_train)
y_pishbini_deci=model_deci.predict(x_test)
```

```python
deghat_deci=r2_score(y_test,y_pishbini_deci)
round(deghat_deci,2)
```

```python
msa_lin=mean_squared_error(y_test,pishbini_model_lin)
msa_lin
```

```python
mae_lin=mean_absolute_error(y_test,pishbini_model_lin)
mae_lin
```

```python
def abzar_alat(y_true,y_pish):
    return{
    "nomre_r2":r2_score(y_true,y_pish),
        "mse":mean_squared_error(y_true,y_pish),
            "mae":mean_absolute_error(y_true,y_pish),


    }
```

```python
pd.DataFrame([abzar_alat(y_test,pishbini_model_lin),abzar_alat(y_test,y_pishbini_deci),abzar_alat(y_test,y_pishbini_svm),abzar_alat(y_test,y_pishbini_kn),abzar_alat(y_test,y_pishbini_random)],index=["lin","deci","SVR","kn","random"])
```

```python
models={
    "Linear Regression": LinearRegression(),
    "KNeighborsRegressor": KNeighborsRegressor(),
    "DecisionTreeRegressor": DecisionTreeRegressor(),
    "SVR": SVR(),
    "Random Forest": RandomForestRegressor()
}
```

```python
behtarin_score=0
behtarin_model=()
for name, model in models.items():
    model.fit(x_train, y_train)
    y_pred=model.predict(x_test)
    score=r2_score(y_test, y_pred)
    if score>behtarin_score:
        behtarin_score=score
        behtarin_model=name
        break
```

```python
print(behtarin_model,behtarin_score)
```

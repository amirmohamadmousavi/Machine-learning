```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.metrics import mean_squared_error,mean_absolute_error
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix,roc_auc_score
```

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "KNN": KNeighborsClassifier(),
    "SVM": SVC(),
    "Random Forest": RandomForestClassifier()
}
```

```python
data=pd.read_csv("breast_cancer_clean.csv")
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model_log=LogisticRegression(max_iter=10000)
model_log.fit(x_train,y_train)
y_pishbini_log=model_log.predict(x_test)
```

```python
deghat_log=accuracy_score(y_test,y_pishbini_log)
round(deghat_log,2)
```

```python
model_svm=SVC()
model_svm.fit(x_train,y_train)
y_pishbini_svm=model_svm.predict(x_test)
```

```python
deghat_svm=accuracy_score(y_test,y_pishbini_svm)#این
round(deghat_svm,2)
```

```python
model_random=RandomForestClassifier(random_state=42)
model_random.fit(x_train,y_train)
y_pishbini_random=model_random.predict(x_test)
```

```python
deghat_random=accuracy_score(y_test,y_pishbini_random)
round(deghat_random,2)
```

```python
model_kn=KNeighborsClassifier()
model_kn.fit(x_train,y_train)
y_pishbini_kn=model_kn.predict(x_test)
```

```python
deghat_kn=accuracy_score(y_test,y_pishbini_kn)
round(deghat_kn,2)
```

```python
model_deci=DecisionTreeClassifier(random_state=42)
model_deci.fit(x_train,y_train)
y_pishbini_deci=model_deci.predict(x_test)
```

```python
deghat_deci=accuracy_score(y_test,y_pishbini_deci)
round(deghat_deci,2)
```

```python
def abzar_alat(y_true,y_pish):
    return{
    "accuracy_score":accuracy_score(y_true,y_pish)
    

    }
```

```python
pd.DataFrame([abzar_alat(y_test,y_pishbini_log),abzar_alat(y_test,y_pishbini_svm),abzar_alat(y_test,y_pishbini_random),abzar_alat(y_test,y_pishbini_kn)],index=["log","SVC","random","kn"])
```

```python
behtarin_accuracy = 0
behtarin_model =()
for name,model in models.items():
    model.fit(x_train, y_train)
    y_pred=model.predict(x_test)
    acc=accuracy_score(y_test, y_pred)
    if acc>behtarin_accuracy:
        behtarin_accuracy=acc
        behtarin_model=name
    if acc==1:
        break
```

```python
print(behtarin_model,behtarin_accuracy)
```
